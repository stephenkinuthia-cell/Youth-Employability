# Model Evaluation
This notebook loads training results and compares the models using evaluation metrics and plots.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid", palette="deep")
```

```python
classification_results = pd.read_csv("classification_model_results.csv")
regression_results = pd.read_csv("regression_model_results.csv")
classification_results
regression_results
```

## Classification comparison

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=classification_results, x="ROC_AUC", y="Model", palette="Blues_r")
plt.title("Classification Model ROC-AUC")
plt.xlabel("ROC-AUC")
plt.tight_layout()
plt.show()
```

```python
classification_long = classification_results.melt(
    id_vars=["Model"],
    value_vars=["Accuracy", "Precision", "Recall", "F1"],
    var_name="Metric",
    value_name="Score"
)
plt.figure(figsize=(10, 6))
sns.barplot(data=classification_long, x="Score", y="Model", hue="Metric")
plt.title("Classification metric comparison")
plt.tight_layout()
plt.show()
```

## Regression comparison

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=regression_results, x="R2", y="Model", palette="Greens_r")
plt.title("Regression Model R2")
plt.xlabel("R2")
plt.tight_layout()
plt.show()
```

```python
regression_long = regression_results.melt(
    id_vars=["Model"],
    value_vars=["RMSE", "MAE"],
    var_name="Metric",
    value_name="Score"
)
plt.figure(figsize=(10, 6))
sns.barplot(data=regression_long, x="Score", y="Model", hue="Metric")
plt.title("Regression error comparison")
plt.tight_layout()
plt.show()
```
